```python
import pandas as pd
data = pd.read_excel("crop yield data sheet.xlsx")
data.head()
```

```python
columns = [data.columns]
for col in columns:
    data[col]=data[col].fillna(data[col].median())
```

```python
data.dtypes
```

```python
data = data[data['Temperatue']!=':']
```

```python
data['Temperatue'].astype(float)
```

```python
from sklearn.model_selection import train_test_split,GridSearchCV

# X = data.drop("Yield (Q/acre)",axis =1)
# y=data["Yield_tons_per_hectare"]

# X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2)
# from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(data.drop('Yeild (Q/acre)', axis = 1), data['Yeild (Q/acre)'], test_size = 0.2, random_state = 42)

len(X_train),len(X_test),len(y_train),len(y_test)
```

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
import numpy as np
np.random.seed(42)
categorical_features = ["Rain Fall (mm)",	"Fertilizer",	"Temperatue",	"Nitrogen (N)",	"Phosphorus (P)",	"Potassium (K)"]

categorical_transform = Pipeline(steps=[
    
    ("onehot",OneHotEncoder(handle_unknown="ignore"))
])

preprocessing = ColumnTransformer(transformers = [
    ("cat",categorical_transform,categorical_features)
]
)

model = Pipeline(steps=[
    ("preprocessor",preprocessing),
    ("model",RandomForestRegressor())
])
model.fit(X_train,y_train)

model.score(X_test,y_test)
```

```python
from sklearn.tree import DecisionTreeRegressor
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
import numpy as np
np.random.seed(42)
categorical_features = ["Rain Fall (mm)",	"Fertilizer",	"Temperatue",	"Nitrogen (N)",	"Phosphorus (P)",	"Potassium (K)"]

categorical_transform = Pipeline(steps=[
    
    ("onehot",OneHotEncoder(handle_unknown="ignore"))
])

preprocessing = ColumnTransformer(transformers = [
    ("cat",categorical_transform,categorical_features)
]
)

model = Pipeline(steps=[
    ("preprocessor",preprocessing),
    ("model",DecisionTreeRegressor())
])
model.fit(X_train,y_train)

model.score(X_test,y_test)
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
import numpy as np
np.random.seed(42)
categorical_features = ["Region","Soil_Type","Crop","Weather_Condition","Fertilizer_Used","Irrigation_Used"]

categorical_transform = Pipeline(steps=[
    
    ("onehot",OneHotEncoder(handle_unknown="ignore"))
])

preprocessing = ColumnTransformer(transformers = [
    ("cat",categorical_transform,categorical_features)
]
)

model = Pipeline(steps=[
    ("preprocessor",preprocessing),
    ("model",LinearRegression())
])
model.fit(X_train,y_train)

model.score(X_test,y_test)
```

```python
from sklearn.tree import DecisionTreeRegressor
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
import numpy as np
np.random.seed(42)
categorical_features = ["Region","Soil_Type","Crop","Weather_Condition","Fertilizer_Used","Irrigation_Used"]

categorical_transform = Pipeline(steps=[
    
    ("onehot",OneHotEncoder(handle_unknown="ignore"))
])

preprocessing = ColumnTransformer(transformers = [
    ("cat",categorical_transform,categorical_features)
]
)

model_dt = Pipeline(steps=[
    ("preprocessor",preprocessing),
    ("model_dt",DecisionTreeRegressor())
])
model_dt.fit(X_train,y_train)

model_dt.score(X_test,y_test)

from sklearn.model_selection import GridSearchCV

#parameters for grid search
para = {
    "model_dt__max_depth": [2,4,6,8],
    "model_dt__min_samples_split": [2,4,6,8],
    "model_dt__min_samples_leaf": [2,4,6,8],
    "model_dt__random_state": [0,42]
}

#Grid search object
grid = GridSearchCV(model_dt, para, cv = 5, n_jobs = -1, verbose = 1)

#fitting the grid search object
grid.fit(X_train, y_train)
```

```python
grid.score(X_test,y_test)

```
